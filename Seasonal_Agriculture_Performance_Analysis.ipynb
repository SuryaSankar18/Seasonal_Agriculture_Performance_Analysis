# **Seasonal Agriculture Performance Analysis**

## **Introduction**

*Agriculture is strongly influenced by seasonal conditions such as rainfall, temperature, humidity, sunlight, soil moisture, irrigation practices, and availability of resources. These factors can directly or indirectly affect crop yield, production, water usage, disease risk, and the overall economic performance of farms.*

*The Seasonal Agriculture Performance Analysis project focuses on examining how agricultural performance varies across different seasons, mainly Kharif, Rabi, and Zaid. The dataset contains information related to farm characteristics, environmental conditions, soil properties, fertilizer and pesticide usage, irrigation methods, crop yield, production, revenue, cost, profit, water consumption, water efficiency, and disease or pest risk.*

*The main purpose of this analysis is to identify meaningful seasonal patterns and understand how environmental conditions, farming practices, resource usage, and economic factors are associated with agricultural outcomes. The analysis also compares crops, irrigation methods, and regional performance to identify strong and weak areas.*

*Using NumPy, Pandas, Matplotlib, and Seaborn, the dataset was cleaned, explored, summarized, and visualized. Missing values and duplicate records were checked, seasonal comparisons were performed, relationships between important variables were examined using correlation, and unusual observations were investigated using plots and descriptive statistics.*

*The overall objective is to generate data-driven insights that can support better agricultural planning, improve resource utilization, identify profitable farming conditions, and understand the major differences between agricultural seasons.*

# **Data Cleaning**

***Check missing values***

```python
missing = df.isnull().sum()

missing[missing > 0]
```

***Missing-value percentage***

```python
missing_report = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": (df.isnull().sum() / len(df)) * 100
})

missing_report[missing_report["Missing Values"] > 0]
```

***Fill missing values using median***

```python
columns_to_fill = [
    "Rainfall_mm",
    "Soil_Moisture_pct",
    "Yield_Tonnes_Ha"
]

for col in columns_to_fill:
    df[col] = df[col].fillna(df[col].median())
```

# **Key Findings**

*The analysis revealed clear differences in agricultural performance across Kharif, Rabi, and Zaid seasons.*

*Kharif emerged as the strongest-performing season overall. It recorded the highest average crop yield at approximately 5.63 tonnes per hectare and also produced the highest total agricultural production, approximately 82,387.61 tonnes. Kharif also generated the highest average revenue and average profit among the three seasons.*

*Average profit during Kharif was approximately ₹178,915 per farm, while Rabi recorded an average profit of approximately ₹87,689. In contrast, Zaid recorded an average loss of approximately ₹24,805 per farm. This indicates that farms operating during Zaid face greater economic challenges compared with those operating during Kharif and Rabi.*

*The environmental analysis showed strong seasonal differences. Kharif received the highest average rainfall at approximately 849.20 mm and also had the highest average humidity and soil moisture. Rabi experienced cooler conditions, while Zaid recorded the highest average temperature at approximately 31.04°C, along with the lowest rainfall and lowest soil moisture.*

*Although Kharif received the most rainfall, the correlation between rainfall and crop yield across the full dataset was only about 0.03. This indicates that rainfall alone has almost no strong linear relationship with yield. Agricultural productivity therefore appears to depend on several interacting factors rather than rainfall alone.*

*Water usage also varied across seasons. Zaid used the highest average quantity of water, approximately 6,419.89 m³ per farm, even though it received the lowest rainfall. However, Zaid also recorded the lowest average water efficiency. Kharif had the highest water efficiency at approximately 5.89 tonnes per 1,000 m³ of water.*

*Irrigation method was associated with noticeable performance differences. Drip irrigation recorded the highest average crop yield at approximately 6.58 tonnes per hectare. However, Rainfed agriculture showed the highest calculated water efficiency, followed by Drip irrigation. This shows that an irrigation method that produces the highest yield may not always be the one with the highest water efficiency.*

*Disease and pest risk also changed across seasons. Kharif recorded the highest average disease and pest risk at approximately 54.47%, compared with lower risk levels during Rabi and Zaid. The combination of higher rainfall, humidity, and soil moisture during Kharif may be associated with these higher risk levels, although the analysis does not establish causation.*

*Crop-level analysis showed large differences in average yield. Sugarcane recorded the highest average yield among the crops in the dataset. However, its yield values are naturally much higher than many other crops, so direct comparisons between fundamentally different crop types should be interpreted carefully.*

*The analysis also showed that crop yield had a strong positive relationship with water efficiency and production. Profit was strongly associated with revenue and also showed positive relationships with production and yield.*

*One of the most unusual patterns in the dataset was observed during Zaid. Despite receiving the lowest rainfall, Zaid used the highest average amount of water and achieved the lowest water efficiency. It also recorded the weakest profitability among the three seasons.*

# **Conclusion**

*The Seasonal Agriculture Performance Analysis demonstrates that agricultural performance differs considerably across seasons. The results show that Kharif performs best overall in terms of average yield, production, revenue, profit, and water efficiency. For example, Drip irrigation is associated with higher average yield, while water efficiency also shows a strong relationship with yield.*

*The economic analysis highlights an important concern regarding Zaid agriculture. Average costs during this season exceed average revenue, resulting in an overall average loss. This suggests that farms operating during Zaid may require better control of input costs, improved water management, or more suitable crop selection.*

*Another important conclusion is that higher resource use does not automatically lead to better performance. Zaid uses the most water on average but has the lowest water efficiency and weakest economic results. Therefore, improving efficiency is likely to be more important than simply increasing resource consumption.*

*Overall, the analysis confirms that seasonal agricultural planning should consider both productivity and efficiency. Decisions should not be based on a single variable such as rainfall, yield, or water usage. Instead, agricultural planning should combine environmental conditions, crop suitability, irrigation practices, resource efficiency, disease risk, and economic performance.*

# **Recommendations**

*Agricultural planning should prioritize season-specific strategies rather than applying the same farming practices across all seasons. Since Kharif, Rabi, and Zaid operate under different environmental conditions, crop selection, irrigation practices, and input use should be adjusted according to the characteristics of each season.*

*Water management should receive particular attention during Zaid. The season uses the highest average quantity of water while recording the lowest water efficiency. Farmers should consider more efficient irrigation practices, better irrigation scheduling, soil-moisture monitoring, and reduction of unnecessary water usage.*

*Drip irrigation should be considered where technically and economically suitable because it is associated with the highest average crop yield in the dataset. However, irrigation decisions should also consider installation cost, crop requirements, local water availability, and overall water efficiency.*

*Special attention should be given to the economic performance of Zaid crops. Since the season shows an average loss, farmers and planners should examine whether high production costs, water usage, crop choice, or market prices are contributing to weak profitability. Crop combinations with better historical profitability should be considered.*

*Kharif farms should strengthen disease and pest management because this season has the highest average disease and pest risk. Preventive monitoring and timely crop protection measures may help reduce potential production losses.*

*Farmers should not depend only on rainfall when making production decisions. Since rainfall has a very weak overall correlation with yield in this dataset, other factors such as soil moisture, crop type, irrigation, seed quality, nutrient availability, and water efficiency should also be considered.*

*Crop-specific analysis should be preferred over comparing all crops only by raw yield. Crops such as sugarcane naturally produce much higher tonnes per hectare than crops such as pulses or cotton, so performance should also be evaluated using profitability, resource efficiency, and suitability for each season.*

*Water efficiency should be monitored as a key performance indicator. The analysis shows a strong relationship between water efficiency and crop yield, making it useful for evaluating whether water resources are being converted effectively into agricultural output.*

*Input costs should be reviewed regularly. Farms with high fertilizer, pesticide, irrigation, or operating expenses should compare these costs with the additional yield and revenue generated to ensure that higher spending is economically justified.*

*Finally, future agricultural planning should combine seasonal performance, crop selection, environmental conditions, water efficiency, disease risk, and economic indicators. Using data-driven analysis can help farmers and agricultural planners improve productivity while controlling costs and conserving resources.*
